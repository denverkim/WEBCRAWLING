```python
# 붓꽃 데이터 열어서 확인
from sklearn import datasets

iris = datasets.load_iris()

print(iris.data)
print(iris.target)
print(iris.feature_names)
print(iris.target_names)
```

```python
# 붓꽃데이터를 열어 타겟과 속성이름을 포함한 df 만들기

import pandas as pd

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target

df.head()
```

```python
# 컬럼이름 변경
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'target']
df.head()
```

```python
# 숫자로 된 범주를 글자로 바꾸기

df.target.replace(0, 'setosa', inplace=True)
df.target.replace(1, 'versicolor', inplace=True)
df.target.replace(2, 'virginica', inplace=True)

df.head()
```

```python
# 데이터프레임 요약

print(df.shape)
print(df.dtypes)
print(df.describe())
df.groupby('target').size()
```

```python
# 데이터 타입 확인
df.dtypes
```

```python
#X and y split. 독립, 종속변수 분리
y = df.target
x = df.drop('target', axis=1)
```

```python
# 레이블 인코더
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(y)
y
```

```python
# 데이터셋 분리
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
#Decision Tree 의사결정트리
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt.fit(x_train, y_train)
dt.score(x_test, y_test) #정확도
```

```python
#Model evaluation (accuracy score) 모델평가
from sklearn.metrics import accuracy_score

y_pred = dt.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
#5 fold cross validation (5 fold) 교차검증

from sklearn.model_selection import cross_val_score

dt_cv = cross_val_score(dt, x_test, y_test, cv=5)
dt_cv
```
