```python
import pandas as pd

golf = pd.read_csv('/content/golf.csv')
golf
```

```python
# Golfing Study Example 골프연구 예제
# Draw a scatter plot. 산점도

import matplotlib.pyplot as plt

plt.scatter(golf.distance, golf.score)
plt.xlim(240, 300)
plt.ylim(0, 100)
plt.xlabel('distance')
plt.ylabel('score')
plt.title('Golf Study')
plt.show()
```

```python
# Calculate the covariance, correlation coefficient. 공분산, 상관관계 계산

import numpy as np

cov = np.cov(golf.distance, golf.score)
corr = np.corrcoef(golf.distance, golf.score)

print('공분산: ', cov[0][1])
print('상관계수: ', corr[0][1])
```

```python
# 판다스를 이용한 공분산, 상관관계 계산
print('공분산')
print(golf.cov())

print('상관관계')
print(golf.corr())
```

```python
import seaborn as sns

sns.heatmap(golf.corr(), annot=True)
plt.show()
```

```python
# Conduct a regression analysis and interpret the results. 회귀분석 및 해석
import statsmodels.api as sm

x = golf.distance
y = golf.score

x = sm.add_constant(x)
model = sm.OLS(y, x)
result = model.fit()

print(result.summary())
```

```python
# 회귀선
sns.lmplot(x='distance', y='score', data=golf, ci=None)
plt.show()
```

```python
sns.regplot(x='distance', y='score', data=golf)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x, y)
print('Y절편: ', model.intercept_)
print('기울기: ', model.coef_[1])
```
